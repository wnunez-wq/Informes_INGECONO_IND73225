```python
import pandas as pd
```

```python
P= 2444.72
i = (1+ 0.0459)**(1/12)-1
n= 360

C = P *(i*(1+i)**n)/ ((1+i)**n-1)
```

```python
saldo = P
datos =[]

for k in range (1, n+1):
    interes = saldo *i
    amortizacion = C- interes 
    saldo_final = saldo - amortizacion
    if k ==n:
        saldo_final =0
    datos.append([k, saldo, interes, amortizacion, C, saldo_final])
    saldo = saldo_final
tabla = pd.DataFrame(datos, columns=["Mes", "Saldo inicial (UF)", "Interés (UF)", "Amortización (UF)", "Cuota (UF)","Saldo final (UF)"])
print("Cuota:", C)
print("\nPrimeros 12 períodos:")
print(tabla.head(12))

print("\nÚltimos 3 períodos:")
print(tabla.tail(3))

print("\nCapital amortizado:", tabla["Amortización (UF)"].sum())
print("Intereses totales:", tabla["Interés (UF)"].sum())
print("Total pagado:", tabla["Cuota (UF)"].sum())
```
