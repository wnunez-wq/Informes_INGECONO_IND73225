```python
import pandas as pd
```

```python

# Pongo el _ en vez de punto para que la cantidad sea legible, al printearlo se muestra igual que usando el punto
montos= [2_000_000, 2_800_000, 3_320_000]
montos
```

```python
ofertas = [ { "Entidad" : "Banco Ripley", "Plazo_dias": 30, "tasa_base_30": 0.0035 } , { "Entidad" : "Banco Ripley", "Plazo_dias": 90, "tasa_base_30": 0.0034}, {"Entidad" : "Banco Falabella", "Plazo_dias":30, "tasa_base_30": 0.0030}, {"Entidad" : "Banco Falabella", "Plazo_dias": 90, "tasa_base_30": 0.0032}, {"Entidad" : "Banco Falabella", "Plazo_dias": 180, "tasa_base_30": 0.0032} ]
ofertas 
```

```python
resultados = []

for oferta in ofertas:
    entidad = oferta["Entidad"]
    plazo = oferta["Plazo_dias"]
    tasa_base_30=oferta["tasa_base_30"]

# tasa correspondiente al plazo completo del deposito
    tasa_periodo = tasa_base_30 *(plazo/30)
    for p in montos:
        F = p * (1+tasa_periodo)
        interes = F-p
        resultados.append({"Entidad": entidad, "Plazo_dias": plazo, "P": p , "Tasa_base_30": tasa_base_30, 
                           "Tasa_periodo": tasa_periodo, "F" : F, "Interes": interes})
```

```python
df = pd.DataFrame(resultados)
```
